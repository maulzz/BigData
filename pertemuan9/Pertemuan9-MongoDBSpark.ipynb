```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["praktikum_bigdata"]

koleksi_produk = db["produk"]

print("Koneksi ke MongoDB Berhasil")
print("Daftar database yang ada:", client.list_database_names())
```

```python
koleksi_produk.delete_many({})

koleksi_produk.insert_one({
    "nama_produk": "Lenovo Legion RTX 3090",
    "kategori": "Laptop",
    "harga": 24000000,
    "stok": 12,
    "tags": ["best-seller", "Gaming"]
})

koleksi_produk.insert_many([
    {"nama_produk": "ASUS ROG Strix G16", "kategori": "Laptop", "harga": 22500000, "stok": 8, "tags": ["Gaming", "High-Performance"]},
    {"nama_produk": "MacBook Air M2", "kategori": "Laptop", "harga": 17000000, "stok": 15, "tags": ["Thin-Light", "Work"]},
    {"nama_produk": "Acer Predator Helios 300", "kategori": "Laptop", "harga": 19500000, "stok": 5, "tags": ["Gaming"]},
    {"nama_produk": "Corsair Vengeance DDR5 32GB (2x16GB)", "kategori": "Komponen", "harga": 1850000, "stok": 25, "tags": ["RAM", "DDR5"]},
    {"nama_produk": "Samsung 980 Pro NVMe SSD 1TB", "kategori": "Komponen", "harga": 1450000, "stok": 30, "tags": ["SSD", "NVMe", "Storage"]}
])

print("Jumlah dokumen di collection 'produk' : ", koleksi_produk.count_documents({}))
```

```python
satu_produk = koleksi_produk.find_one({"kategori": "Laptop"})
print("Satu produk kategori Minuman:", satu_produk)

print()

print("Seluruh produk kategori Laptop:")
for p in koleksi_produk.find({"kategori": "Laptop"}):
    print(f"  - {p['nama_produk']} (Rp{p['harga']:,})")

print()

print("Produk dengan harga > 20000000(20 Juta):")
for p in koleksi_produk.find({"harga": {"$gt": 20000000}}):
    print(f"  - {p['nama_produk']} (Rp{p['harga']:,})")
```

```python
koleksi_produk.update_one(
    {"nama_produk": "MacBook Air M2"},
    {"$set": {"harga": 21000000}}
)

koleksi_produk.update_many(
    {"nama_produk": "MacBook Air M2"},
    {"$inc": {"stok": -1}}
)

print("Setelah Update")
for p in koleksi_produk.find({}, {"_id": 0, "nama_produk": 1, "harga": 1, "stok": 1}):
    print(f"  - {p['nama_produk']}: harga=Rp{p['harga']:,}, stok={p['stok']}")
```

```python
koleksi_produk.delete_one({"nama_produk": "ASUS ROG Strix G16"})

print("Jumlah dokumen tersisah: ", koleksi_produk.count_documents({}))
print("Daftar produk tersisah: ", [p["nama_produk"] for p in koleksi_produk.find({}, {"nama_produk": 1, "_id": 0})])
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, count
import pandas as pd

spark = SparkSession.builder.appName("Pertemuan9-MongoDBSpark").master("local[*]") .getOrCreate()
```

```python
dokumen_mongo = list(koleksi_produk.find({}))

df_pandas = pd.DataFrame(dokumen_mongo)

df_pandas["_id"] = df_pandas["_id"].astype(str)
df_pandas["tags"] = df_pandas["tags"].apply(lambda x: ", ".join(x) if isinstance(x, list) else "")

df_spark_produk = spark.createDataFrame(df_pandas)

df_spark_produk.printSchema()
df_spark_produk.show()
```

```python
df_spark_produk.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
    count("nama_produk").alias("jumlah_produk")
).show()

```

```python
spark.stop()
print("SparkSession ditutup.")
```
